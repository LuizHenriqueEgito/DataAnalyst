```python
import pandas as pd
import altair as alt
```

```python
df = pd.DataFrame({
    "x": [1, 2, 3, 4],
    "y": [10, 20, 30, 40],
    "z": [1, 0, 1, 0]
})
print(df)
```

```python
plot = alt.Chart(df).mark_point().encode(x='x', y='y', color=alt.Color('z', scale=alt.Scale(scheme='reds')))
```
