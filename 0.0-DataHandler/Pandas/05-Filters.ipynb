````python
import pandas as pd

from pathlib import Path
````

# Filtros Booleanas
São filtros baseados em `True` e `False` onde ele traz apenas quem é `True`

````python
mask = df['credit_score'] > 600  # Coloca quem é True
df[mask].head()
````

É possivel colocar filtros compostos utilizando `& (e)`, `| (ou)` e `~ (negação)`.

````python
# utilizando E (&)
filter_1 = (df['credit_score'] > 600) & (df['geography'] == 'Spain')
df[filter_1].head()
````

````python
# utilizando OU (|)
filter_2 = (df['credit_score'] > 600) | (df['geography'] == 'Spain')
df[filter_2].head()
````

````python
df.card_type.value_counts()
````

````python
# utilizando o isin
filter_3 = df['card_type'].isin(['DIAMOND', 'GOLD'])
df[filter_3].head()
````

````python
# utilizando a negação (~)
filter_4 = df['card_type'].isin(['DIAMOND', 'GOLD'])
df[~filter_4].head()
````

````python
# podemos utilizar o between
# inclusive: {'both', 'neither', 'left', 'right'}
df['credit_score'].between(200, 600, inclusive='both')
````

````python
# Podemos encadear vários filtros
filter_1 = df['point_earned'] > 200
filter_2 = df['estimated_salary'] < 10_000
filter_3 = df['credit_score'].between(200, 600, inclusive='both')
filter_4 = df['card_type'].isin(['DIAMOND', 'GOLD'])

df[filter_1 & filter_2 & filter_3 & filter_4].head()
````

````python
# ou podemos fazer de uma vez
df[
    (df['point_earned'] > 200) &
    (df['estimated_salary'] < 10_000) &
    (df['credit_score'].between(200, 600, inclusive='both')) &
    (df['card_type'].isin(['DIAMOND', 'GOLD']))
].head()
````

# `.loc[]` e `.iloc[]`
`loc` - Seleção por **rótulos** (labels)
- Usa os nomes das linhase e colunas;
- Inclui o último elemento do intervalo;
```python
df.loc[rows (index), cols (nomes das colunas)]
```
Você pode fazer assim:
```python
df.loc[index_0: indext_5, coluna_d: coluna_z]
```

`iloc` - Seleção por **posições** (indices inteiros)
- Usa as posições numéricas;
- Exclui o último elemento do intervalo (como slicing no Python)
```python
df.iloc[0: 11, 2: 5]  # pela até a 10º linha e da 2º até a 4º coluna
```

````python
df.loc[0: 5, 'credit_score': 'gender']
````

````python
df.iloc[0: 6, 2: 5]
````

````python
cols = ['surname', 'geography', 'balance']  # colunas salteadas
df.loc[1:5, cols]
````

# `query()`
Query se assemelha muito ao `SQL` usando expressões em `string`.
```python
# filtro simples
df.query('col_a > 10')

# filtros multiplos
df.query('col_a > 10 and col_b < 500')
df.query('col_a > 10 & col_b < 500')

# usando variáveis externas
value = 10
df.query('col_a > @value')
list_values = ['a', 'b', 'c']
df.query('col_d in @list_values')
# negação
df.query('col_d not in @list_values')

# operações com string
df.query('col_a.str.constains("exp")')
df.query('col_a.str.startswith("abc")')
df.query('col_a.str.endswith("abc")')
df.query('col_a.str.len() > 5')

# valores nulos
df.query('col_a.isna()')  # df.query('idade != idade')
df.query('idade.notna()')  # df.query('idade == idade')

# filtrando pelo index
df.query('index > 2')
df.query('index in [0, 2, 4]')
```

# Valores Faltantes (NA's)

````python
df = pd.DataFrame({
    'col_a': [1, 2, None, 4, 5],
    'col_b': ['x', None, 'z', None, 'y'],
    'col_c': [1, 2, 3, 4, None],
})
df
````

# Linhas Duplicadas

````python
df = pd.DataFrame({
    "A": [1, 1, 2, 3, 3, 3],
    "B": ["x", "x", "y", "z", "z", "z"],
})
df
````

````python
# Mostrando todas as linhas duplicadas
df.duplicated()
````

````python
# Perceba nos indices
df[df.duplicated()]  # use keep por default: df[df.duplicated(keep='first')]
````

````python
# Deixa apenas o último
df[df.duplicated(keep='last')]
````

````python
# Pegando apenas os duplicados
df[df.duplicated(keep=False)]
````

````python
# Removendo as duplicadas
# podemos usar o keep, como 'first', 'last' ou False
# se for False REMOVE todas as linhas duplicadas e deixa apenas as que não tem duplicadas
df.drop_duplicates()
````

# Tempo de execução


Concluimos que:
- Entre 0 e 10.000.000 a filtragem padrão com `.values` é o método mais rápido (usa numpy);
- Entre 0 e 20.000 `query` é o método mais lento;
- Depois de 10.000.000 `np.where` é o método mais rápido.

## Comparativo

````python
import numpy as np

N = 100_000_000  # 100 milhões
data = {
    'cat': np.random.choice(['foo', 'bar', 'baz'], size=(N)),
    'num': np.random.rand(N)
}
df_big = pd.DataFrame(data)
````

````python
# filtro normal
%timeit df_big[(df_big["cat"] == "bar") & (df_big["num"] < 0.5)]
````

````python
%timeit df_big[(df_big["cat"].values == "bar") & (df_big["num"].values < 0.5)]
````

````python
%timeit df_big.query("cat == 'bar' and num < 0.5")
````

# Summary
Utilize a filtragem booleana padrão com `.values` e caso mesmo assim esteja ruim a performance migre para o `polars`.
