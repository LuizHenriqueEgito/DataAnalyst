# Introdução
Aqui nos conseguimos configurar alguns parametros de como queremos que seja feito o funcionamento do pandas. Por exemplo podemos colocar quantas `linhas`, `colunas` ver por padrão ao rodar algo como:
```python
df
```
Podemos colocar também para que esse display na tela arredonde as colunas float para ser facil a visualização mas sem alterar os valores das colunas, é apenas para melhor visualização.
Por exemplo temos uma coluna com valor *0.93932421* mas queremos olhar ela através do Display no `jupyter` podemos configurar o display para arredondar esse numero em 3 casas decimais e veriamos algo como *0.939*. Mas ao fazer contas com esse valor ele ainda usaria o valor padrão dele *0.93932421*.

Todos essas configurações podem ser feitas olhando as opções que o **pandas** nos permite mexer:
```python
import pandas as pd

pd.describe_option()
```

````python
import pandas as pd
import numpy as np
````

````python
pd.describe_option()
````

````python
df = pd.DataFrame(
    np.random.rand(100, 25),
    columns=[f"col_{i+1}" for i in range(25)]
)
````

````python
# Ele mostra 10 linhas e 20 colunas (10 primeiras e 10 ultimas) por padrão
df
````

````python
pd.set_option("display.max_rows", 4)  # qtde de linhas mostradas (o limite é 10)
pd.set_option("display.max_columns", 4)  # qtde de colunas mostradas
pd.set_option("display.float_format", "{:.1f}".format)  # display das colunas float (não altera o número em sí apenas a visualização)
````

````python
df
````

````python
pd.set_option("display.max_rows", None)  # mostra todas as linhas
pd.set_option("display.max_columns", None)  # mostra todas as colunas
````

````python
df
````

````python
pd.reset_option("all")  # volta tudo para a configuração que padrão
````

````python
df
````
