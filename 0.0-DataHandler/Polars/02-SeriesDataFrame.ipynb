# O que é uma `Series`
É um `Dataframe` com uma unica `coluna` **(1D)** tipada em que é possivel fazer operações vetorizadas com ele.
- `DataFrame`: tabela **2D**
- `Series`: coluna **1D**

Um `Dataframe` são várias `Series` organizadas.

```python
# pegando apenas uma Coluna
print(df['CreditScore'].head())
```

```python
type(df['CreditScore'])
```

```python
s = df['CreditScore']
```

```python
# podemos fazer operações vetoriais em uma Series
(s * 3).mean()
```

```python
# podemos filtrar uma Series
print(s.filter(s >= 800))
```

```python
# perceba que usando .select em uma coluna só temos um dataframe
type(df.select('CreditScore'))
```
