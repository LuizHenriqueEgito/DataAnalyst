# Como formatar valores númericos (EXCEL)

- #.##..." B"
- #.##.." M"
- #.##." K"
- #.##0

## 1. O que signigica `#` e `0` nessa formatação:
- `#`: Mostra dígitos se eles existirem, não mostra zeros
```txt
    Exemplo:
    Número: 12
    Formato: ### → 12
    Número: 0 → em branco
```
- `0`: Sempre mostra um dígito, coloca 0 se o número não tiver aquele dígito
```txt
    Exemplo:
    Número: 12
    Formato: 000 → 012
```
## 2. Ponto e vírgula (;):
O `Excel` usa até 4 seções de formatação
```txt
    PORITIVO ; NEGATIVO ; ZERO ; TEXTO
```
Assim podemos fazer:
```txt
    #,##0" M";[Vermelho]-#,##0" M";"Nada"
```
Isso nos daria:
- Positivos: `#,##0" M"`
- Negativos: `[Vermelho]-#,##0" M"`
- Zero: `Nada` 

## 3. O ponto(.) na formatação:
Quando temos:
```txt
    #.##
```
O `Excel` mostra até 2 casas decimais (ou quantas você escrever)

## 4. Aspas duplas:
É o texto que colocamos na nossa formatação
