```python
import pandas as pd
from pathlib import Path
```

```python
FILE_PATH = Path('../../data/customer_churn_records.csv')
df = pd.read_csv(FILE_PATH)
```

# O que é uma `Series`
É um `Dataframe` com uma unica `coluna` **(1D)** tipada em que é possivel fazer operações vetorizadas com ele.
- `DataFrame`: tabela **2D**
- `Series`: coluna **1D**

Um `Dataframe` são várias `Series` organizadas.
No `Pandas` uma `Series` possui `index` para funcionar como se fosse um dicionario ordenado, não só uma lista de valores **1D**.

```python
s = df['CreditScore']  # ou simplesmente: df.CreditScore
s
```

```python
type(s)
```

```python
# com o index do pandas você pode acessar o valor
s[0]
```

```python
# podemos fazer operações vetoriais em uma Series
(s * 3).mean()
```

```python
# podemos filtrar uma Series
s[s > 800]
```
