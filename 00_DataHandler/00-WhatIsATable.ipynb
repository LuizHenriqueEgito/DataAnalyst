# O que é uma tabela?
Uma tabela é uma forma de `tabular` de organizar informações. É como uma grade, onde cada casela guarda um pedaço de informação. Ela é composta por `linhas` e `colunas`, por exemplo:
```txt
+----------+--------+--------+
|  Nome    | Idade  | Cidade |
+----------+--------+--------+
| Ana      |   24   | Recife |
| João     |   31   | Santos |
| Maria    |   29   | Natal  |
+----------+--------+--------+
```
Essa estrutura é uma tabela.

## Colunas
Cada coluna representa um **tipo de informação** dessa forma todas as células de uma coluna guardam o mesmo tipo de dado.

## Linhas
Cada linha representa **um registro**, uma entidade completa.

## Summary (Tabelas são)
- Fáceis de ler;
- Fáceis de organizar;
- Fáceis de fazer buscas;
- Fáceis de serem salvas;
- Fáceis de manipular (fazer filtros e transformações)

Em suma podemos encarar uma `tabela` como uma `matriz` com linhas e colunas e podemos fazer operações `vetoriais` com elas:


Podemos chamar uma tabela de `DataFrame`.

Além disso podemos encarar cada coluna separadamenteé o que chamamos de `Series`:


````python
import pandas as pd
````

````python
col_peso = [60.2, 70.1, 81.2, 55.1]
col_altura = [1.72, 1.85, 1.90, 1.60]
col_idade = [21, 22, 18, 27]
````

````python
dataframe = pd.DataFrame(
    data={
        'Peso': col_peso,
        'Altura': col_altura,
        'Idade': col_idade
    }
)
dataframe
````
