````python
import re
import sqlite3
import pandas as pd
from pathlib import Path
````

````python
FILE_DATASET = Path('data/CustomerChurnRecords.csv')
````

````python
df = pd.read_csv(FILE_DATASET)
````

Baixe e instale o SQLLite: [Aqui](https://sqlitebrowser.org/dl/)

The first thing we're going to do to deal with our table is to **standardize** the names of our columns.

````python
df = df.set_index('RowNumber')
````

````python
def standardize_cols(col_name: str) -> str:
    snake = re.sub(r'\s+', '', col_name)
    snake = re.sub(r'(?<!^)(?=[A-Z])', '_', snake).lower()
    return snake
````

````python
df.columns = [standardize_cols(col) for col in df.columns]
````

````python
TABLE_NAME = 'customer_churn_records'
````

````python
with sqlite3.connect("databse.db") as conn:
    df.to_sql(TABLE_NAME, conn, if_exists="replace", index=False)
````
